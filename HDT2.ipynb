## RNN a la atencion


1. El cuello de botella en el modelo encode-decoder con RNN se da por su funcionamiento, ya que intenta meter toda la información en un vector de tamaño fijo y hace que un solo estado lleve toda la carga. Empeora con las oraciones largas por la pérdida de memoria ya que la RNN no logra sostener tanta informacion a la larga y no olvidar que el vector es de tamaño fijo, por lo que no puede guardar tantos detalles. Al tner una gran cantidad de información el decodificador no recibe datos en buen estado.
La atencion de Bahdanau lo resuelve por medio de eliminar la necesidad de meter toda la entrada en el vector.

2. La diferencia entre estas dos es la matematica que usan para calcular los pesos entre los estados ocultos del codificador y decodificador. Aunque Luong es mas eficiente en hardware moderno debido a que esta basado en multiplicaciones de matrices optimizadas.

3. La analogia cae en que en un diccionario se guarda infomracion en forma de "llave:valor" y la busqueda suave se hace uso de query para comprar que tan parecidas son las keys con esta y sed les asigna un peso segun que tan parecidas/relevantes sean.

4. Las limitaciones que se eliminan son quitar la recurrencia y lo susituye por el mecanismo de auto atencion, hace que ahora el flujo de trabajo sea paralelo.
